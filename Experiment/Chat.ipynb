```python
import os
import dotenv
dotenv.load_dotenv(override=True)
```

```python
google_api_key = os.getenv("GOOGLE_API_KEY")
if not google_api_key:
    raise ValueError("GOOGLE_API_KEY is not set in the environment variables.")
else:
    print("GOOGLE_API_KEY is set successfully.")
```

```python
from llama_index.core import SimpleDirectoryReader #Role --> Data Loading --> how? Reads documents from a directory and converts them into LlamaIndex Document objects. Depending on installed file readers, it can handle files such as TXT, PDF, DOCX, and CSV.
from llama_index.core import VectorStoreIndex # Role --> Indexing & Semantic Retrieval --> how? 1- docs --> chunks.  2- chunks --> embeddings.  3- embeddings --> vector store.  4- vector store --> index 5- Finds relevant chunks when a question is asked.
from llama_index.llms.gemini import Gemini # Role --> Response generation and reasoning.
from IPython.display import Markdown,display # Role --> Displaying results in a Jupyter notebook.
from llama_index.core import ServiceContext # Role --> Configuring the LLM and other components.(Group them together to create a service context that can be used to initialize the index and query engine.)
from llama_index.core import StorageContext # Role --> Managing persistent index data.
from llama_index.core import load_index_from_storage # Role --> Loading an existing index.
import google.generativeai as genai
from llama_index.embeddings.gemini import GeminiEmbedding # Role -->Converting documents and questions into vectors for semantic search.
from llama_index.core import Settings # Role --> Modern configuration for the LLM, embedding model, chunking, and related defaults.
```

```python
genai.configure(api_key=google_api_key) 
```

```python
for model in genai.list_models():
    print(model)    
    
```

```python
for model in genai.list_models():
    if "generateContent" in model.supported_generation_methods:
        print(f"Model Name: {model.name}")
```

```python
Data = SimpleDirectoryReader("../Data")
```

```python
Data = Data.load_data()
```

```python
print(Data[0].text)
```

```python
model = Gemini(models="gemini-2.0-flash", temperature=0.2, max_output_tokens=1024 , api_key=google_api_key)
```

```python
gemini_embed_model = GeminiEmbedding(
    model_name="models/gemini-embedding-001",
    api_key=google_api_key,
)
```

```python
#Settings for LLM, embedding model, chunking, and related defaults.
Settings.llm = model
Settings.embed_model = gemini_embed_model
Settings.chunk_size = 400
Settings.chunk_overlap = 10
```

```python
index = VectorStoreIndex.from_documents(Data)
```

```python
query_engine = index.as_query_engine()
```
